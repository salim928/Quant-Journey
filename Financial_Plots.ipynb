```python
import pandas as pd
import plotly.express as px
```

```python
import plotly.graph_objects as go
```

```python
raw = pd.read_csv("/content/fxcm_eur_usd_eod_data.csv", index_col=0, parse_dates=True)
```

```python
raw.info()
```

*  Selects four columns from the DataFrame object (Open-High-Low-Close, or
OHLC).
* Only a few data rows are used for the visualization.
* Returns the final five rows of the resulting data set quotes.

```python
raw.head(5)
```

```python
quotes = raw[["OpenAsk","HighAsk","LowAsk","CloseAsk"]]
quotes = quotes.iloc[-60:]
quotes.tail()
```

```python
quotes['Date'] = quotes.index
```

```python
fig = go.Figure(data=[
    go.Candlestick(
        x = quotes['Date'],
        open=quotes['OpenAsk'],
        high=quotes['HighAsk'],
        low=quotes['LowAsk'],
        close=quotes['CloseAsk']

    )
])
# Final layout
fig.update_layout(
    title='Candlestick Chart',
    xaxis_title='Date',
    yaxis_title='Price',
    xaxis_rangeslider_visible=False
)

fig.show()
```

```python
# Calculate 20-day Simple Moving Average and Bollinger Bands
window = 20
quotes['SMA'] = quotes['CloseAsk'].rolling(window=window).mean()
quotes['STD'] = quotes['CloseAsk'].rolling(window=window).std()
quotes['Upper Band'] = quotes['SMA'] + (2 * quotes['STD'])
quotes['Lower Band'] = quotes['SMA'] - (2 * quotes['STD'])

# Create candlestick chart
fig = go.Figure()

fig.add_trace(go.Candlestick(
    x=quotes['Date'],
    open=quotes['OpenAsk'],
    high=quotes['HighAsk'],
    low=quotes['LowAsk'],
    close=quotes['CloseAsk'],
    name='Candlestick'
))

# Add SMA and Bollinger Bands
fig.add_trace(go.Scatter(
    x=quotes['Date'],
    y=quotes['SMA'],
    line=dict(color='blue', width=1),
    name='SMA (20)'
))

fig.add_trace(go.Scatter(
    x=quotes['Date'],
    y=quotes['Upper Band'],
    line=dict(color='green', width=1, dash='dot'),
    name='Upper Band'
))

fig.add_trace(go.Scatter(
    x=quotes['Date'],
    y=quotes['Lower Band'],
    line=dict(color='red', width=1, dash='dot'),
    name='Lower Band'
))

# Final layout
fig.update_layout(
    title='Candlestick Chart with Bollinger Bands',
    xaxis_title='Date',
    yaxis_title='Price',
    xaxis_rangeslider_visible=False
)

fig.show()
```
